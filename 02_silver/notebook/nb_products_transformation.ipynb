## Products table transformation

```python
data = spark.read.table("`01_bronze`.raw.products") 
display(data) 
```

## Removing anything rather than number in price column

```python
from pyspark.sql import functions as F

def normalize_prizes(df, col_names):
    for col in col_names:
        df = df.withColumn(
            col, 
            F.regexp_replace(F.col(col), '[^0-9.]', '').cast('double')
        )
    return df

data = normalize_prizes(data, ['unit_cost_usd', 'unit_price_usd'])
display(data)
```

## Casting the datatype's of the column and save as processed tables

```python
data = data.withColumn("productkey", F.trim(F.col("productkey")).cast('int'))

data.write.mode('overwrite').saveAsTable("`02_silver`.transformation.products")
```
