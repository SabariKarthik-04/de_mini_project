## Reading the exchange rates table from raw table

```python
data = spark.read.table("`01_bronze`.raw.exchange_rates")
display(data)
```

## Date Normalization to standard format


```python
import pyspark.sql.functions as F

def standardize_date(columnName, df):
    return df.withColumn(
        "date_parsed",
        F.coalesce(
            F.try_to_date(F.col(columnName), "M/d/yyyy"),
            F.try_to_date(F.col(columnName), "M-d-yyyy"),
            F.try_to_date(F.col(columnName), "yyyy-M-d"),
            F.try_to_date(F.col(columnName), "yyyy/M/d")
        )
    ).withColumn(
        columnName,
        F.trim(F.col("date_parsed")).cast('date')
    ).drop("date_parsed")

data = standardize_date("date", data)
display(data)

```

## Casting the exchange column datatype to double

```python
data = data.withColumn("exchange", F.col("exchange").cast("double"))
display(data)
```

## writing data to the silver data lake table

```python
data.write.mode("overwrite").saveAsTable("02_silver.transformation.exchange_rates")
```
